# Bloodline 2.0: Mule Network & Money DNA Exploratory Analysis
This notebook demonstrates exploratory graph mining, rapid pass-through detection, next-hop probability calibration, and comparative evaluation against rule-only baselines on disclosed synthetic benchmark data.

```python
import networkx as nx
import pandas as pd
import numpy as np
from app.db.store import BloodlineDatastore
from app.services.graph_service import GraphService
from app.services.lineage_service import LineageService
from app.services.prediction_service import NextHopPredictionService

store = BloodlineDatastore.get_instance()
print(f"Loaded {len(store.accounts)} accounts and {len(store.transactions)} transactions in multi-layer graph.")
```

```python
# 1. Graph Structural Feature Extraction
graph_svc = GraphService(store)
mule_metrics = graph_svc.detect_mule_patterns("ACC-MULE-HUB1")
print("Mule Hub Patterns:", mule_metrics)
```

```python
# 2. Money DNA Lineage Flow
lineage_svc = LineageService(store)
dna_response = lineage_svc.build_money_dna("TXN-ORIGIN-75K")
print(f"Origin: ${dna_response.origin_amount} -> Tracked ${dna_response.total_tainted_tracked} across {len(dna_response.nodes)} nodes.")
```

```python
# 3. Next-Hop Ranking & Interception
pred_svc = NextHopPredictionService(store)
predictions = pred_svc.predict_next_hop("ACC-MULE-002", top_k=3)
for c in predictions.candidates:
    print(f"Rank #{c.rank}: {c.account_id} | Prob: {c.predicted_probability:.3f} | Action: {c.action_proposal}")
```
